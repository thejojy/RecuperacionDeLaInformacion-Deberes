```python
import pandas as pd

# Cargar el archivo CSV
ruta_archivo = '/content/drive/MyDrive/Universidad/Recuperacion informacion/Examen 1b/V1/Data/rotten_tomatoes_movies.csv'
df_peliculas = pd.read_csv(ruta_archivo)

# Mostrar las primeras filas para verificar la carga
print(df_peliculas.head())
```

##2.- Preprocesamiento

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Definir funciones de procesamiento
def limpiar_caracteres(texto):
    if texto is None:
        return None
    return re.sub(r"[^a-zA-Z\s]", "", texto)  # Solo dejar letras y espacios

def eliminar_stopwords(texto, stop_words):
    if texto is None:
        return None
    palabras = texto.split()  # Dividir el texto en palabras
    palabras_filtradas = [palabra for palabra in palabras if palabra not in stop_words]
    return " ".join(palabras_filtradas)

def aplicar_stemming(texto, stemmer):
    if texto is None:
        return None
    palabras = texto.split()  # Divide el texto en palabras
    return " ".join([stemmer.stem(palabra) for palabra in palabras])  # Reconstruye el texto

# Configuración de preprocesamiento
stop_words = set(stopwords.words('english'))  # Lista de stopwords en inglés
stemmer = PorterStemmer()  # Crear el stemmer

# Preprocesar datos
df["Texto_procesado"] = df["review_content"].fillna("").str.lower()  # Convertir a minúsculas y manejar valores nulos
df["Texto_procesado"] = df["Texto_procesado"].apply(limpiar_caracteres)  # Limpiar caracteres
df["Texto_procesado"] = df["Texto_procesado"].apply(lambda x: eliminar_stopwords(x, stop_words))  # Eliminar stopwords
df["Texto_procesado"] = df["Texto_procesado"].apply(lambda x: aplicar_stemming(x, stemmer))  # Aplicar stemming

# Mostrar las primeras filas del DataFrame
print(df.head())
```

##3.- Vectorizacion

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Inicializar el vectorizador
vectorizador = TfidfVectorizer(max_features=1000, ngram_range=(1, 2), min_df=3)

# Generar la matriz TF-IDF
tfidf_matriz = vectorizador.fit_transform(df["Texto_procesado"])

# Mostrar el vocabulario
print("Vocabulario:", vectorizador.get_feature_names_out())

# Mostrar la matriz TF-IDF
print("Matriz TF-IDF:")
print(tfidf_matriz.toarray())

```

##4.- Construccion de la busqueda

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd


# Función para procesar una consulta
def procesar_consulta(consulta, stop_words, stemmer):
    consulta = consulta.lower()  # Convertir a minúsculas
    consulta = limpiar_caracteres(consulta)  # Limpiar caracteres
    consulta = eliminar_stopwords(consulta, stop_words)  # Eliminar stopwords
    consulta = aplicar_stemming(consulta, stemmer)  # Aplicar stemming
    return consulta

#Funcion para la busqueda de la consulta
def buscar(consulta_usuario):

  # Procesar la consulta
  consulta_procesada = procesar_consulta(consulta_usuario, stop_words, stemmer)

  # Vectorizar la consulta procesada utilizando el mismo vectorizador
  consulta_vectorizada = vectorizador.transform([consulta_procesada])

  # Calcular similitudes entre la consulta y las reseñas
  similitudes = cosine_similarity(consulta_vectorizada, tfidf_matriz)

  # Crear un DataFrame temporal con las similitudes
  df_similitudes = pd.DataFrame({
    "rotten_tomatoes_link": df["rotten_tomatoes_link"],
    "critic_name": df["critic_name"],
    "review_type": df["review_type"],
    "review_score": df["review_score"],
    "review_date": df["review_date"],
    "review_content": df["review_content"],
    "similitud": similitudes[0]
  })

  # Filtrar por similitudes mayores o iguales al 70% (0.7)
  df_filtrado = df_similitudes[df_similitudes["similitud"] >= 0.7]

  # Ordenar por similitud descendente
  df_filtrado = df_filtrado.sort_values(by="similitud", ascending=False)

  if df_filtrado.empty:
    print("No se encontraron reseñas con similitud >= 70%.")
    return None

  df_combinado = pd.merge(
    df_filtrado,
    df_peliculas,
    on="rotten_tomatoes_link",
    how="left"  # Si alguna reseña no tiene una película asociada, usa "left" para mantener las reseñas.
  )

  columnas_relevantes = [
    "similitud", "critic_name", "review_date", "review_content", "review_type", "review_score",
    "movie_title", "genres", "directors", "actors",
    "original_release_date", "runtime", "production_company",
    "critics_consensus", "tomatometer_rating", "audience_rating"
  ]

  df_final = df_combinado[columnas_relevantes]
  return df_final

def imprimir_resultados(df_final, consulta):
  print("Para la busqueda ->"+ consulta +"<- se dio los siguientes resultados")

  if df_final is not None:
    for _, row in df_final.iterrows():
        print(f"Similitud: {row['similitud']:.2f}")
        print(f"Crítico: {row['critic_name']}")
        print(f"Fecha de la reseña: {row['review_date']}")
        print(f"Reseña: {row['review_content']}")
        print(f"Tipo de reseña: {row['review_type']} | Puntuación: {row['review_score']}")
        print(f"Película: {row['movie_title']}")
        print(f"Género: {row['genres']}")
        print(f"Directores: {row['directors']}")
        print(f"Actores: {row['actors']}")
        print(f"Fecha de estreno: {row['original_release_date']} | Duración: {row['runtime']}")
        print(f"Compañía productora: {row['production_company']}")
        print(f"Consenso de críticos: {row['critics_consensus']}")
        print(f"Tomatometer: {row['tomatometer_rating']}% | Audiencia: {row['audience_rating']}%")
        print("-" * 50)
  else:
    print("No se encontraron reseñas relevantes con similitud >= 70%.")

```

##5.- Simulacion de consultas

```python
# Ejemplo de consulta de usuario
consulta_usuario = "Movie with great acting and thrilling plot!"

# Buscar la consulta
df_final = buscar(consulta_usuario)

# Imprimir los resultados
imprimir_resultados(df_final, consulta_usuario)

```
